### Importing modules in python
- Modules and packages in python
- They help to organize and reuse code.

```python
import math
```

```python
math.sqrt(8)
```

```python
from math import sqrt,pi
print(sqrt(16))
print(pi)
```

```python
import numpy as np
l=np.array([1,2,3,4])
type(l)
```

```python
## imports all the function from the math module
from math import *
print(sqrt(16))
```

```python
math.exp(7)
```

This is used to import the modules that we created in the same directory itself, and we are importing the functions from the modules and using them in our code.

```python
from packages.maths import addition,subtraction
```

```python
addition(8,7)
```

```python
subtraction(8,3)
```

```python
from packages.subpackages.mul import multiply
multiply(8,6)
```
