### Custom exceptions
- They are user-defined error classes that you create to represent specific errors in your program instead of using only built-in exceptions.


```python
class Error(Exception):
    pass
class dobException(Error):
    pass
```

- They make code: more readable, easier to debug, easier to handle specific application errors.

```python
year=int(input("Enter the dob:"))
age = 2025-year
try:
    if age<=30 and age>=20:
        print("The age is valid, you cam apply")
    else:
        raise dobException
except dobException:
    print("Sorry, you are not eligible")
```

```python
class InsufficientBalanceError(Exception):
    
    def __init__(self, balance, withdraw_amount):
        self.balance = balance
        self.withdraw_amount = withdraw_amount
        
        super().__init__(
            f"Balance {balance} is insufficient for withdrawing {withdraw_amount}"
        )

try:
    balance = 1000
    withdraw = 1500

    if withdraw > balance:
        raise InsufficientBalanceError(balance, withdraw)

except InsufficientBalanceError as e:
    print(e)
```

```python
class DogNameNotFoundError(Exception):
    def __init__(self,name):
        self.name=name

        super().__init__(
            f'Dog name "{name}" not found in the database'
        )

try:
        dog_name = (input("Enter the dog's name: ")).title()
        dog_database = ["Buddy", "Max", "Charlie"]

        if dog_name not in dog_database:
            raise DogNameNotFoundError(dog_name)

        print(f"Hello, {dog_name}!")
        
except DogNameNotFoundError as e:
        print(e)
```
