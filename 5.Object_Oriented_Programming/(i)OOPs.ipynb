### Classes and Objects
- Object-Oriented Programming (OOP) is a programming paradigm that uses "objects" to design applications and computer programs. 
- OOP allows for modelling real-world scenarios using classes and abjects. 
- This lesson covers the basics of creating classes and objects.

A class is a blueprint for creating objects.
- It has attributes (data) and methods (functions) that define the behavior of the objects created from the class.

An object is an instance of a class.
- It is created using the class and can have its own unique attributes and methods.

```python
class Car:
    pass
audi=Car()
bmw = Car()
```

##### Instance Variables and Methods
Instance variables are attributes that belong to an instance of a class. 
- They are defined within the __init__ method and are unique to each object created from the class.
- Self keyword is used to help access the instance variable inside the class.

```python
class Dogs:
    ## constructor
    def __init__(self,name,age):
        self.name=name ## instance variable
        self.age=age

## here, name and age are the property for the dog class or attributes
## There can be multiple methods in a class as well as multiple variables.

## creat objects
dog1=Dogs("bunny",3)

print(dog1)
print(dog1.name)
print(dog1.age)
```

```python
dog2=Dogs("Lucy",3)
```

Defining a class with instance methods

```python
## define a clas with instance methods
class Dog:
    def __init__(self, name, age):
        self.name=name
        self.age=age
    # def __init__(self, birthdate):
    #     self.birthdate=birthdate
    ## this overwrites the previous init method and now we can only create object.
    def bark(self):
        print(f"{self.name} says woof")
```

```python
dog1=Dog("bunny",3)
print(dog1.name)
```

```python
## modelling a bank account
# define a class for bank account
owners=[]

class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner=owner
        self.balance=balance
    
    def owners_list(self):
        owners.append(self.owner)
        return owners
    
    def delete_account(self,owner):
        if owner in owners:
            owners.remove(owner)
            print(f"{owner} account is deleted")
        else:
            print("Account not found!")

    def deposit(self,amount):
        self.balance+=amount
        print(f"{amount} is deposited and account balance is now {self.balance}")

    def withdraw(self, amount):
        if amount>self.balance:
            print("Insufficient funds!")
        else:
            self.balance-=amount
            print(f"{amount} is withdrawn. new balance is {self.balance}")
            
    def get_balance(self):
        return self.balance
```

```python
## deposit
account1=BankAccount("Alice",balance=10000)
```

```python
account1.deposit(5000)
account1.get_balance()
```

```python
account1.withdraw(3000)
```

```python
account1.delete_account('Alice')
```
