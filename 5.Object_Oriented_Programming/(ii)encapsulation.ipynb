### Encapsulation and Abstraction
These two are the two fundamental principles of OOPS that help in desiging robust, maintainable, and reusable code. 
- **Encapsulation**: It involves bundling data and methods that operate on the data within a single unit.
- **Abstraction**: It involves hiding complex implementation details and exposing only necessary features.

#### Encapsulation
- It is a concept of wrapping data(variables) and methods(functions) together as a single unit. 
- It restricts direct access to some of the object's components, which is a means of preventing accidental interference and misuse of the data.

```python
## Encapsulation with Getter and Setter Methods
## Public, protected and private variables or access modifiers.
class Person:
    def __init__(self,name, age):
        self.name=name ## public variables
        self.age=age ## public variables
def get_name(person):
    return person.name
person=Person("John", 30)
print(person.name) # Output: John
```

```python
get_name(person) # Output: John
```

```python
dir(person)
```

```python
## private variables
class Person:
    def __init__(self,name, age,gender):
        self.__name=name ## private variables
        self.__age=age ## private variables
        self.gender=gender ## public variables
```

We cannot access the private variable directly outside the class, even in the derived class. We can access it using getter and setter methods.

Protected variables are those that are intended to be accessed only within the class and its subclasses. In Python, we can indicate a variable as protected by prefixing it with a single underscore (_). However, this is just a convention and does not enforce any access restrictions.

```python
## protected variables
class Person:
    def __init__(self,name, age,gender):
        self._name=name ## protected variables
        self._age=age ## protected variables
        self.gender=gender ## public variables
class Employee(Person):
    def __init__(self,name,age,gender):
        super().__init__(name,age,gender)
    
emp=Employee('krish',34,'Male')
print(emp._name) # This will print "krish"
```

#### Encapsulation with Getter and Setter methods

```python
class Dataperson:
    def __init__(self,name, age):
        self.__name=name ## private variables
        self.__age=age ## private variables
    ## getter and setter methods
    def get_name(self):
        return self.__name
    def set_name(self, name):
        self.__name=name
    def get_age(self):
        return self.__age
    def set_age(self, age):
        if age>0:
          self.__age=age
        else:
          print("Age must be positive and greater than zero")
```

##### Accessing private variables using getter and setter methods

```python
person1=Dataperson('John',30)
```

```python
person1.get_name() # Output: John
```

```python
person1.set_name("Jane")
```
