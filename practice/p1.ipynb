```python
active=['lara','manon','lisa','rose','jisoo','jennie']
```

```python
active
```

```python
len(active)
```

```python
inactive_list=[]
```

```python
n=input("Enter named account you want to delete:")
f=0
for i in active[:]:
    if n==i:
        f=1
        inactive_list.append(i)
        active.pop(active.index(i))

if(f==1):
    print(f"User found and account deleted with name:{n}")
else:
    print("User not found and you must create an account before deleting it.")

print("\n")

print("Number of active accounts are:")
for i in active:
    print(i,end=" ")

print("\n")

print("Number of inactive accounts are:")
for i in inactive_list:
    print(i,end=" ")
```

```python
n=input("Enter named account you want to delete:")
f=0
for i in active[:]:
    if n==i:
        f=1
        inactive_list.append(i)
        active.remove(n)

if(f==1):
    print(f"User found and account deleted with name:{n}")
else:
    print("User not found and you must create an account before deleting it.")

print("\n")

print("Number of active accounts are:")
for i in active:
    print(i,end=" ")

print("\n")

print("Number of inactive accounts are:")
for i in inactive_list:
    print(i,end=" ")
```
