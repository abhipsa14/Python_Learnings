```python
def city(city,country):
    return f"{city},{country}"

cit=input("Enter your city:")
country=input("Enter your country:")


print(city(cit,country))
```

```python
def album(artist,album_name,no_of_songs=None):
    dictionary={'artist':artist,'album_name':album_name,'no_of_songs':no_of_songs}
    return dictionary

while(True):
    print("Choice you have:")
    print("1. Start")
    print("2. Continue")
    print("3. Quit")
    ch=int(input("Your choice:"))
    print(f"Your choice is {ch}")

    if ch==1:
        artist=input("Enter your favourite artist name:")
        album_name=input("Enter album name:")
        no_of_songs=input("No. of songs:")
        print(album(artist,album_name,no_of_songs))
    elif ch==2:
        ch=1
    else:
        break

```

```python
def short_messages(messages,sent_messages):
    while messages:
        message=messages.pop()
        print(message)
        sent_messages.append(message)
def sent_message(sent_messages):
    while sent_messages:
        print(sent_messages.pop(),end=" ")


messages=["hi","hello","How are you?"]
sent_messages=[]
short_messages(messages[:],sent_messages)
sent_message(sent_messages)
```
