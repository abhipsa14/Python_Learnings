## Python:
- It is case sensitive language

```python
print("Hello World!")
message="Hello World!"
print(message)
# message='Bobby's World!'
message="Bobby's World!"
print(message)
```

```python
message="""Hi, I am Abhipsa. I am learning Python.
I am loving it."""
print(message)
```

```python
len(message)
message=message.upper()
message=message.lower()
message[9:16]
message.count('i')
message.find('abhipsa')
print(message)
new_message=message.replace('abhipsa','universe')
new_message
```
